#  Интерполяционные сплайны

### 📌 Определение 1 Сплайн 

**Сплайном** $S_m(x)$ называется определенная на отрезке $[a, b]$ функция, принадлежащая классу $C^l[a, b]$ ($l$ раз непрерывно дифференцируемых функций), такая, что на каждом промежутке $[x_{k-1}, x_k]$ ($k = 1, 2, \dots, n$) она является многочленом $m$-й степени.

Разность между степенью сплайна $m$ и показателем его гладкости $l$ называется **дефектом сплайна**:

$$d := m - l$$

## Известные сплайны

### 📌 Определение 2 Кубический сплайн дефекта 1

**Кубическим сплайном дефекта 1**, интерполирующим на отрезке $[a, b]$ данную функцию $f(x)$, называется функция $g(x)$, которая на каждом промежутке $[x_{k-1}, x_k]$ ($k = 1, 2, \dots, n$) представляется в виде многочлена:

$$g_k(x) := a_k + b_k(x - x_k) + c_k(x - x_k)^2 + d_k(x - x_k)^3 \qquad \text{(11.8)}$$

и удовлетворяет следующей совокупности условий:

* **а) Условие интерполяции в узлах сплайна:**
  $$g(x_k) = f(x_k) \quad (k = 0, 1, \dots, n)$$

* **б) Двойная непрерывная дифференцируемость (гладкость):**
  $$g(x) \in C^2[a, b]$$
  *(Это значит, что в точках стыка $x_k$ совпадают значения самих сплайнов, их первых и вторых производных).*

* **в) Краевые (граничные) условия:**
  $$g''(a) = g''(b) = 0$$
  *(Сплайны с такими условиями называют **естественными**).*

### 📜 Теорема. Существование и единственность кубического сплайна

При заданных на концах отрезка $[a, b]$ краевых условиях:
$$g''(a) = g''(b) = 0$$

по заданным в узлах сетки $a = x_0, x_1, x_2, \dots, x_{n-1}, x_n = b$ значениям $f_k = f(x_k)$ ($k = 0, 1, \dots, n$) функции $f(x)$ можно построить **единственный** интерполирующий её кубический сплайн $g(x)$ дефекта $1$.

## Базисные сплайны

### 📌 Определение 3 Базисный сплайн (B-сплайн) степени $m-1$

**Базисным сплайном** или, короче, **B-сплайном** степени $m - 1$ (дефекта 1) относительно узлов $x_k < x_{k+1} < \dots < x_{k+m}$ называется функция:

$$B_{m-1, k}(x) = B_{m-1}(x_k; x_{k+1}; \dots ; x_{k+m}; x) := m \sum_{i=k}^{k+m} \frac{(x_i - x)_+^{m-1}}{\Pi_{m, k}(x_i)} \qquad \text{}$$

где:

$$(x_i - x)_+^{m-1} := [\max\{0; (x_i - x)\}]^{m-1} \qquad \text{}$$

$$\Pi_{m, k}(x) := (x - x_k)(x - x_{k+1}) \dots (x - x_{k+m}) \qquad \text{}$$
