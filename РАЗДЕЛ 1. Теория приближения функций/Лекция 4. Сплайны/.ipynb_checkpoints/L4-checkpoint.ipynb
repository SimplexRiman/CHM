# Лекция 4. 
#  Интерполяционные сплайны

### 📌 Определение 1 Сплайн 

**Сплайном** $S_m(x)$ называется определенная на отрезке $[a, b]$ функция, принадлежащая классу $C^l[a, b]$ ($l$ раз непрерывно дифференцируемых функций), такая, что на каждом промежутке $[x_{k-1}, x_k]$ ($k = 1, 2, \dots, n$) она является многочленом $m$-й степени.

Разность между степенью сплайна $m$ и показателем его гладкости $l$ называется **дефектом сплайна**:

$$d := m - l$$

---

## Известные сплайны

### 📌 Определение 2 Кубический сплайн дефекта 1

**Кубическим сплайном дефекта 1**, интерполирующим на отрезке $[a, b]$ данную функцию $f(x)$, называется функция $g(x)$, которая на каждом промежутке $[x_{k-1}, x_k]$ ($k = 1, 2, \dots, n$) представляется в виде многочлена:

$$g_k(x) := a_k + b_k(x - x_k) + c_k(x - x_k)^2 + d_k(x - x_k)^3 \qquad \text{(11.8)}$$

и удовлетворяет следующей совокупности условий:

* **а) Условие интерполяции в узлах сплайна:**
  $$g(x_k) = f(x_k) \quad (k = 0, 1, \dots, n)$$

* **б) Двойная непрерывная дифференцируемость (гладкость):**
  $$g(x) \in C^2[a, b]$$
  *(Это значит, что в точках стыка $x_k$ совпадают значения самих сплайнов, их первых и вторых производных).*

* **в) Краевые (граничные) условия:**
  $$g''(a) = g''(b) = 0$$
  *(Сплайны с такими условиями называют **естественными**).*

### 📜 Теорема. Существование и единственность кубического сплайна

При заданных на концах отрезка $[a, b]$ краевых условиях:
$$g''(a) = g''(b) = 0$$

по заданным в узлах сетки $a = x_0, x_1, x_2, \dots, x_{n-1}, x_n = b$ значениям $f_k = f(x_k)$ ($k = 0, 1, \dots, n$) функции $f(x)$ можно построить **единственный** интерполирующий её кубический сплайн $g(x)$ дефекта $1$.

## Базисные сплайны

### 📌 Определение 3 Базисный сплайн (B-сплайн) степени $m-1$

**Базисным сплайном** или, короче, **B-сплайном** степени $m - 1$ (дефекта 1) относительно узлов $x_k < x_{k+1} < \dots < x_{k+m}$ называется функция:

$$B_{m-1, k}(x) = B_{m-1}(x_k; x_{k+1}; \dots ; x_{k+m}; x) := m \sum_{i=k}^{k+m} \frac{(x_i - x)_+^{m-1}}{\Pi_{m, k}'(x_i)} \qquad \text{}$$

где:

$$(x_i - x)_+^{m-1} := [\max\{0; (x_i - x)\}]^{m-1} \qquad \text{}$$

$$\Pi_{m, k}(x) := (x - x_k)(x - x_{k+1}) \dots (x - x_{k+m}) \qquad \text{}$$

### Рекуррентная формула Кокса — де Бора 

Пусть задана последовательность узлов $x_k < x_{k+1} < \dots < x_{k+m}$. Для вычисления B-сплайна $B_{m-1, k}(x)$ степени $m-1$ используется рекуррентный процесс по шагам степени $p$ (от $p=0$ до $p=m-1$):

**Шаг 0: Базис нулевой степени ($p = 0$)**
$$
B_{0, k}(x) = 
\begin{cases} 
  1, & \text{если } x_k \le x < x_{k+1} \\ 
  0, & \text{иначе} 
\end{cases}
$$

**Шаг $p$: Рекурсивный переход для степени $p > 0$**
$$
B_{p, k}(x) = \dfrac{x - x_k}{x_{k+p} - x_k} B_{p-1, k}(x) + \dfrac{x_{k+p+1} - x}{x_{k+p+1} - x_{k+1}} B_{p-1, k+1}(x)
$$

Финальный искомый сплайн степени $m-1$ получится на последнем шаге и будет равен $B_{m-1, k}(x)$.

> **Правило раскрытия неопределенностей:** Если в знаменателе любой из дробей возникает ноль (при совпадении узлов), вся эта дробь целиком считается равной $0$.

## Пространства спалйнов

### 📜 Теорема.  Пространство сплайнов степени $m$ дефекта 1

#### Формулировка
Множество всех сплайнов $s(x)$ степени $m$ дефекта $1$, заданных на сетке узлов $\Delta: a = x_0 < x_1 < \dots < x_n = b$, образует **векторное (линейное) пространство** над полем вещественных чисел. 

Это пространство обозначается как $S_m(\Delta)$.


### 📜 Теорема. О размерности и локальном базисе из B-сплайнов (Теорема Карри - Шёнберга)

#### Формулировка теоремы
Пусть на отрезке $[a, b]$ задана сетка из $n$ интервалов $\Delta: a = x_0 < x_1 < \dots < x_n = b$. 
Дополним её $2m$ произвольными граничными узлами так, чтобы:
$$
x_{-m} \le x_{-(m-1)} \le \dots \le x_{-1} \le x_0 = a \quad \text{и} \quad b = x_n \le x_{n+1} \le \dots \le x_{n+m}
$$

Тогда:
1. Размерность пространства сплайнов $S_m(\Delta)$ степени $m$ дефекта $1$ равна:
$$
\dim S_m(\Delta) = n + m
$$
2. Система B-сплайнов $\{B_{m, k}(x)\}_{k=-m}^{n-1}$ степени $m$, построенных на этой расширенной сетке узлов, является **линейно независимой** на отрезке $[a, b]$ и образует **базис** пространства $S_m(\Delta)$.

Любой сплайн $s(x) \in S_m(\Delta)$ на отрезке $[a, b]$ однозначно разлагается по этому базису:
$$
s(x) = \sum_{k=-m}^{n-1} C_k B_{m, k}(x)
$$
