# Лекция 3. 
# Многочлены Чебышёва

### 📌 Определение 1.1
**Многочленом Чебышева*** называется функция:

$$T_n (x) := \cos(n \arccos x) \qquad \text{(1.1)}$$

где $n \in \mathbb{N}_0$, а $x \in [-1, 1]$.

> ℹ️ **Историческая справка:** 
> *Чебышев Пафнутий Львович (1821–1894)* — знаменитый русский математик. Его работы о многочленах наилучшего равномерного приближения легли в основу конструктивной теории функций. Стандартное обозначение $T_n(x)$ происходит от немецкого написания фамилии — *Tschebyschew*.

---



### 📊 Вывод рекуррентной формулы

Прежде всего, убедимся, что функция $T_n(x)$, представленная с помощью тригонометрических функций, на самом деле является многочленом при любом $n = 0, 1, 2, \dots$

1. **Базовые значения (непосредственная подстановка):**
   При $n = 0$ и $n = 1$ получаем:
   $$T_0(x) = 1$$
   $$T_1(x) = x$$

2. **Связь через тригонометрию:**
   Положим $\alpha := \arccos x$. Тогда:
   * $T_1(x) = \cos \alpha$
   * $T_n(x) = \cos n\alpha$
   * $T_{n-1}(x) = \cos(n - 1)\alpha$
   * $T_{n+1}(x) = \cos(n + 1)\alpha$

3. **Использование формулы суммы косинусов:**
   Так как $\cos(n + 1)\alpha + \cos(n - 1)\alpha = 2 \cos \alpha \cos n\alpha$, справедливо равенство:
   $$T_{n+1}(x) + T_{n-1}(x) = 2 T_1(x) T_n(x)$$

4. **Рекуррентное соотношение:**
   Перепишем полученное равенство в удобном виде:
   $$\mathbf{T_{n+1}(x) = 2x T_n(x) - T_{n-1}(x)} \qquad \text{(1.2)}$$
   
   Данная формула определяет последовательность функций при $n = 1, 2, 3, \dots$, где $x \in [-1, 1]$.

---



### 📈 Первые шесть многочленов Чебышева

Подставляя начальные члены $T_0(x)$ и $T_1(x)$ в рекуррентную формулу, находим последующие элементы последовательности:

* **$T_0(x)$** $= 1$
* **$T_1(x)$** $= x$
* **$T_2(x)$** $= 2x^2 - 1$
* **$T_3(x)$** $= 2x(2x^2 - 1) - x = 4x^3 - 3x$
* **$T_4(x)$** $= 2x(4x^3 - 3x) - (2x^2 - 1) = 8x^4 - 8x^2 + 1$
* **$T_5(x)$** $= 2x(8x^4 - 8x^2 + 1) - (4x^3 - 3x) = 16x^5 - 20x^3 + 5x$

### 📐 Свойство 1.1. Корни многочленов Чебышёва

Многочлен Чебышёва $T_n(x)$ имеет на отрезке $[-1, 1]$ ровно $n$ различных действительных корней. Все они лежат строго внутри интервала $(-1, 1)$ и задаются формулой:

$$x_k = \cos \left( \frac{2k + 1}{2n} \pi \right) \qquad \text{}$$

где $k = 0, 1, 2, \dots, n - 1$.


### 📐 Свойство 1.2. Перемежаемость корней и экстремумов (Альтернанс)

Корни многочленов Чебышёва **перемежаются** (строго чередуются) с точками их наибольших и наименьших значений, которые равны соответственно $+1$ и $-1$.

Для многочлена $T_n(x)$ на отрезке $[-1, 1]$ экстремумы принимают значения:
$$T_n(x_j) = (-1)^j$$

Они достигаются в точках:
$$x_j = \cos \left( \frac{j}{n} \pi \right) \qquad \text{}$$

где $j = 0, 1, 2, \dots, n$.

---

### 📐 Свойство 1.3 (Теорема Чебышёва о наименьшем уклонении)

Из всех многочленов степени $n$ со старшим коэффициентом, равным $1$, нормированный (приведенный) многочлен Чебышёва $\tilde{T}_n(x)$ **наименее уклоняется от нуля** на отрезке $[-1, 1]$.

Математически это означает, что для любого другого многочлена $P_n(x) = x^n + a_{n-1}x^{n-1} + \dots + a_0$ справедливо неравенство:

$$\max_{x \in [-1, 1]} |P_n(x)| \ge \max_{x \in [-1, 1]} |\tilde{T}_n(x)| = \frac{1}{2^{n-1}}$$

## О многочленах наилучших равномерных приближений

### 📜 Теорема (Теорема Чебышёва о чебышёвском альтернансе)

Многочлен $Q_n(x)$ степени $n$ является **многочленом наилучшего равномерного приближения** для непрерывной функции $f(x) \in C[a, b]$ тогда и только тогда, когда на отрезке $[a, b]$ существует не менее $n + 2$ точек $x_j$, в которых разность между функцией и многочленом поочередно достигает своего максимума с разными знаками.

То есть в точках $x_1 < x_2 < \dots < x_{n+2}$ выполняется условие:
$$f(x_j) - Q_n(x_j) = \pm (-1)^j \cdot E$$

Где $E$ — это максимальное уклонение многочлена от функции на всём отрезке:
$$E := \|f(x) - Q_n(x)\|_{C[a, b]} = \max_{x \in [a, b]} |f(x) - Q_n(x)|$$
