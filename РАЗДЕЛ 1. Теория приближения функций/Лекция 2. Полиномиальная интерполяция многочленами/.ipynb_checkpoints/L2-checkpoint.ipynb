# Лекция 2. 
# Полиномиальная интерполяция многочленами

# 1 Основные определения

---
### 1.1 Аппроксимация

Пусть:

- $X$ — **метрическое пространство** с расстоянием $d(\cdot, \cdot)$ (например, пространство непрерывных функций $C[a,b]$ с равномерной метрикой, или гильбертово пространство $L_2[a,b]$ со среднеквадратичной нормой).
- $f \in X$ — исходный объект (функция, набор данных).
- $M \subset X$ — **множество допустимых приближений** (например, многочлены степени не выше $n$, тригонометрические полиномы, сплайны и т.д.).
---

### Определение
---
**Аппроксимацией** элемента $f$ элементами множества $M$ называется задача нахождения элемента (или последовательности элементов) $g \in M$, для которого выполняется:

$$
d(f, g) = \inf_{h \in M} d(f, h)
$$

Если точная нижняя грань достигается, то $g$ называется **элементом наилучшего приближения**.

Если же точная нижняя грань не достигается, или нас устраивает приближённое решение, то говорят об **аппроксимации с заданной точностью $\varepsilon > 0$**:

$$
\exists g \in M : \quad d(f, g) \le \varepsilon.
$$

---

### Частные метрики (виды аппроксимации)

| Вид аппроксимации | Метрика (норма) | Условие минимизации |
|-------------------|----------------|----------------------|
| **Равномерная (чебышёвская)** | $\displaystyle d(f,g) = \max_{x \in [a,b]} |f(x)-g(x)|$ | Минимизация максимального отклонения |
| **Среднеквадратичная ($L_2$)** | $\displaystyle d(f,g) = \left( \int_a^b |f(x)-g(x)|^2 dx \right)^{1/2}$ | Минимизация интеграла квадрата разности |
| **Дискретная (МНК)** | $\displaystyle d(f,g) = \sum_{i=1}^n (y_i - g(x_i))^2$ | Минимизация суммы квадратов невязок в точках |

### Ключевая особенность
Аппроксимация **не требует** совпадения в отдельных точках; она минимизирует глобальную меру отклонения на всей области.


### 1.2 Интерполяция

#### Исходные данные
---
Заданы:

- $n+1$ попарно различных **узлов** $x_0, x_1, \dots, x_n$ (обычно из отрезка $[a,b]$).
- Значения $y_i = f(x_i)$ или произвольные числа в этих узлах.
- Класс функций $\Phi$ (например, многочлены степени $\le n$, сплайны, тригонометрические полиномы).

---

### Определение
---
**Интерполяцией** называется построение функции $g \in \Phi$ такой, что выполняются **условия точного совпадения** во всех заданных узлах:

$$
g(x_i) = y_i, \qquad i = 0, 1, \dots, n.
$$

При этом поведение $g$ между узлами не регламентируется дополнительными условиями (кроме тех, что накладываются самим классом $\Phi$, например, гладкость).

### Важное замечание
Интерполяция является **частным случаем аппроксимации** с нулевой погрешностью в узлах, но **не обязана** минимизировать какую-либо интегральную или равномерную норму. Между узлами интерполяционный многочлен может сильно отклоняться от исходной функции (эффект Рунге).

---

### 1.3 Экстраполяция
---
Пусть:

- Известны значения функции $f$ (или набор точек $(x_i, y_i)$) только на некотором интервале $[a,b]$.
- На этом интервале построена некоторая модель $g(x)$ (например, с помощью интерполяции или аппроксимации).
---

### Определение
---
**Экстраполяцией** называется вычисление значения $g(x)$ для аргумента $x$, **не принадлежащего** исходному интервалу $[a,b]$, т.е. для $x < a$ или $x > b$:

$$
f(x) \approx g(x) \quad \text{при} \quad x \notin [a,b].
$$

Иными словами, экстраполяция — это **продолжение** зависимости за пределы известной области.

---

# 2 Интерполяционный многочлен Лагранжа

Пусть функция $y = f(x)$ задана в виде таблицы своих значений в $n+1$ различных узлах:

| $x$   | $x_0$ | $x_1$ | ... | $x_n$ |
|-------|-------|-------|-----|-------|
| $y$   | $y_0$ | $y_1$ | ... | $y_n$ |

где $y_i = f(x_i)$.

### 2.1. Основная идея

Будем искать многочлен $L_n(x)$ степени не выше $n$, удовлетворяющий условиям:

$$
L_n(x_i) = y_i = f(x_i), \quad i = 0, 1, \ldots, n
$$



### 2.2. Формула Лагранжа

Интерполяционный многочлен Лагранжа записывается в виде:

$$
L_n(x) = \sum_{i=0}^{n} y_i \cdot \ell_i(x)
$$

где $\ell_i(x)$ — **базисные многочлены Лагранжа**, определяемые формулой:

$$
\ell_i(x) = \prod_{\substack{j=0 \\ j \neq i}}^{n} \frac{x - x_j}{x_i - x_j}
$$


В развёрнутом виде:

$$
\ell_i(x) = \frac{(x - x_0)(x - x_1)\cdots(x - x_{i-1})(x - x_{i+1})\cdots(x - x_n)}
{(x_i - x_0)(x_i - x_1)\cdots(x_i - x_{i-1})(x_i - x_{i+1})\cdots(x_i - x_n)}
$$

Каждый базисный многочлен $\ell_i(x)$ имеет степень $n$ и обладает **ключевым свойством**:

$$
\ell_i(x_j) = 
\begin{cases}
1, & i = j \\
0, & i \neq j
\end{cases}
= \delta_{ij}
$$

Это свойство гарантирует, что построенный многочлен $L_n(x)$ проходит через все заданные точки:

$$
L_n(x_i) = \sum_{j=0}^{n} y_j \cdot \ell_j(x_i) = y_i
$$


### 2.3. Частные случаи

#### Линейная интерполяция ($n = 1$)

Для двух точек $(x_0, y_0)$ и $(x_1, y_1)$:

$$
L_1(x) = \frac{x - x_1}{x_0 - x_1} y_0 + \frac{x - x_0}{x_1 - x_0} y_1
$$

Это уравнение прямой, проходящей через две заданные точки.

#### Квадратичная интерполяция ($n = 2$)

Для трёх точек $(x_0, y_0)$, $(x_1, y_1)$, $(x_2, y_2)$:

$$
L_2(x) = \frac{(x - x_1)(x - x_2)}{(x_0 - x_1)(x_0 - x_2)} y_0 +
\frac{(x - x_0)(x - x_2)}{(x_1 - x_0)(x_1 - x_2)} y_1 +
\frac{(x - x_0)(x - x_1)}{(x_2 - x_0)(x_2 - x_1)} y_2
$$

Это парабола, проходящая через три точки.

---


### 2.4 Пример построения многочлена Лагранжа

**Пример.** Построить интерполяционный многочлен Лагранжа для функции, заданной таблицей:

| $x$ | 2 | 3 | 4 |
|-----|---|---|---|
| $y = \ln x$ | $\ln 2$ | $\ln 3$ | $\ln 4$ |

Для $n = 2$ имеем:

$$
L_2(x) = \frac{(x - 3)(x - 4)}{(2 - 3)(2 - 4)} \ln 2 +
\frac{(x - 2)(x - 4)}{(3 - 2)(3 - 4)} \ln 3 +
\frac{(x - 2)(x - 3)}{(4 - 2)(4 - 3)} \ln 4
$$

$$
L_2(x) = \frac{(x - 3)(x - 4)}{2} \ln 2 -
(x - 2)(x - 4) \ln 3 +
\frac{(x - 2)(x - 3)}{2} \ln 4
$$

Чтобы найти приближённое значение функции в точке $x^*$, не совпадающей с узлами, достаточно подставить $x^*$ в $L_2(x)$.

### 2.5 Единственность интерполяционного многочлена

**Теорема.** Интерполяционный многочлен степени не выше $n$, проходящий через $n+1$ различных точек, **единственен**.

**Доказательство.** Предположим, что существуют два многочлена $p_n(x)$ и $q_n(x)$ степени не выше $n$, удовлетворяющие условиям интерполяции. Их разность $d(x) = p_n(x) - q_n(x)$ также является многочленом степени не выше $n$ и имеет $n+1$ различных корней $x_0, x_1, \ldots, x_n$. Но многочлен степени $n$ не может иметь более $n$ корней, если только он не является тождественным нулём. Следовательно, $d(x) \equiv 0$, то есть $p_n(x) \equiv q_n(x)$.


---

### 2.6 Погрешность интерполяции

####  Остаточный член

Если функция $f(x)$ имеет $(n+1)$-ю производную на отрезке $[a, b]$, содержащем все узлы, то погрешность интерполяции оценивается формулой:

$$
R_n(x) = f(x) - L_n(x) = \frac{f^{(n+1)}(\xi)}{(n+1)!} \cdot \omega_{n+1}(x)
$$

где $\xi$ — некоторая точка из интервала $(a, b)$, а

$$
\omega_{n+1}(x) = (x - x_0)(x - x_1)\cdots(x - x_n)
$$



#### Практическая оценка погрешности

На практике часто используют оценку:

$$
|f(x) - L_n(x)| \leq \frac{M_{n+1}}{(n+1)!} \cdot |\omega_{n+1}(x)|
$$

где $M_{n+1} = \max_{x \in [a,b]} |f^{(n+1)}(x)|$.

Важно отметить, что высокая степень интерполяционного многочлена не всегда гарантирует малую погрешность — это связано с **явлением Рунге**.

# 3. Конечные разности

### 3.1. Определение и основные понятия

Пусть задана функция $y = f(x)$, значения которой известны в равноотстоящих узлах:

$$
x_i = x_0 + i h, \qquad i = 0, 1, 2, \ldots
$$

где $h > 0$ — **шаг** (постоянное расстояние между соседними узлами).  
Обозначим соответствующие значения функции:

$$
y_i = f(x_i), \qquad i = 0, 1, 2, \ldots
$$


**Конечной разностью первого порядка** (или первой разностью) называется разность между двумя соседними значениями функции:

$$
\Delta y_i = y_{i+1} - y_i, \qquad i = 0, 1, 2, \ldots
$$

Здесь символ $\Delta$ — **оператор конечной разности** (оператор сдвига вперёд).

---

### 3.2 Конечные разности высших порядков

Конечные разности **второго порядка** определяются через разности первого порядка:

$$
\Delta^2 y_i = \Delta (\Delta y_i) = \Delta y_{i+1} - \Delta y_i = y_{i+2} - 2y_{i+1} + y_i
$$

Аналогично, разности **третьего порядка**:

$$
\Delta^3 y_i = \Delta^2 y_{i+1} - \Delta^2 y_i
$$

В общем случае, **разность $k$-го порядка** определяется рекуррентно:

$$
\Delta^k y_i = \Delta^{k-1} y_{i+1} - \Delta^{k-1} y_i, \qquad k = 1, 2, 3, \ldots
$$

причём $\Delta^0 y_i = y_i$.

#### Явная формула для разности $k$-го порядка

Разность $k$-го порядка можно выразить через значения функции в узлах с помощью биномиальных коэффициентов:

$$
\Delta^k y_i = \sum_{j=0}^{k} (-1)^j \binom{k}{j} y_{i+k-j}
$$

или, что эквивалентно:

$$
\Delta^k y_i = y_{i+k} - k y_{i+k-1} + \frac{k(k-1)}{2!} y_{i+k-2} - \ldots + (-1)^k y_i
$$

---

### 3.3 Основные свойства оператора конечной разности $\Delta$

1. **Линейность**:  
   Для любых функций $u(x)$ и $v(x)$ и произвольной постоянной $C$:
   $$
   \Delta (C u(x)) = C \Delta u(x)
   $$
   $$
   \Delta (u(x) + v(x)) = \Delta u(x) + \Delta v(x)
   $$

2. **Разность от постоянной равна нулю**:  
   $$
   \Delta C = 0
   $$

3. **Разность от многочлена**:  
   Если $P_n(x)$ — многочлен степени $n$, то его первая разность — многочлен степени $n-1$, вторая — степени $n-2$, и так далее.  
   Разность $n$-го порядка от многочлена степени $n$ есть **константа**:
   $$
   \Delta^n P_n(x) = a_0 n! h^n
   $$
   где $a_0$ — старший коэффициент многочлена.  
   Все разности порядка выше $n$ равны нулю.

4. **Связь с производной**:  
   При малом шаге $h$ конечная разность приближает производную:
   $$
   \Delta y_i \approx h \cdot f'(x_i) + O(h^2)
   $$
   Точнее:
   $$
   f'(x_i) \approx \frac{\Delta y_i}{h}
   $$

---


### 3.4 Таблица конечных разностей (диагональная схема)

Для наглядного представления конечных разностей удобно использовать таблицу, где значения функции и все их разности располагаются по диагонали.

| $x$ | $y$ | $\Delta y$ | $\Delta^2 y$ | $\Delta^3 y$ | $\dots$ |
|-----|-----|------------|--------------|--------------|---------|
| $x_0$ | $y_0$ | $\Delta y_0$ | $\Delta^2 y_0$ | $\Delta^3 y_0$ | ... |
| $x_1$ | $y_1$ | $\Delta y_1$ | $\Delta^2 y_1$ | ... | |
| $x_2$ | $y_2$ | $\Delta y_2$ | ... | | |
| $x_3$ | $y_3$ | ... | | | |
| ... | ... | | | |


**Пример заполнения** (для первых строк):

- $\Delta y_0 = y_1 - y_0$
- $\Delta y_1 = y_2 - y_1$
- $\Delta^2 y_0 = \Delta y_1 - \Delta y_0 = y_2 - 2y_1 + y_0$
- $\Delta^3 y_0 = \Delta^2 y_1 - \Delta^2 y_0$

---

### 3.5 Пример вычисления конечных разностей

**Пример.** Дана таблица значений функции $y = \ln x$ при $x = 2, 3, 4, 5$ с шагом $h = 1$:

| $x$ | 2 | 3 | 4 | 5 |
|-----|---|---|---|---|
| $y$ | 0.6931 | 1.0986 | 1.3863 | 1.6094 |

Построим таблицу конечных разностей до второго порядка:

| $x$ | $y$ | $\Delta y$ | $\Delta^2 y$ |
|-----|-----|------------|--------------|
| 2   | 0.6931 | 0.4055 | -0.1178 |
| 3   | 1.0986 | 0.2877 | -0.0646|
| 4   | 1.3863 | 0.2231 | |
| 5   | 1.6094 | | |

**Вычисления:**

- $\Delta y_0 = 1.0986 - 0.6931 = 0.4055$
- $\Delta y_1 = 1.3863 - 1.0986 = 0.2877$
- $\Delta y_2 = 1.6094 - 1.3863 = 0.2231$
- $\Delta^2 y_0 = \Delta y_1 - \Delta y_0 = 0.2877 - 0.4055 = -0.1178$
- $\Delta^2 y_1 = \Delta y_2 - \Delta y_1 = 0.2231 - 0.2877 = -0.0646$


---


# 4. Конечноразностные интерполяционные формулы

Пусть для функции $y = f(x)$ заданы значения $y_0, y_1, \ldots, y_n$ в равноотстоящих узлах $x_0, x_1, \ldots, x_n$ с шагом $h > 0$, т.е. $x_i = x_0 + ih$, где $i = 0, 1, \ldots, n$.

Задача состоит в том, чтобы построить интерполяционный многочлен $P_n(x)$ степени не выше $n$, удовлетворяющий условиям $P_n(x_i) = y_i$.

Для решения этой задачи воспользуемся аппаратом конечных разностей.

Будем искать многочлен $P_n(x)$ в виде, обобщающем разложение функции в ряд Тейлора, но с использованием конечных разностей вместо производных. Пусть $x = x_0 + qh$, где $q$ — некоторое число. Тогда, как показано в § 8.4, имеет место следующее представление.

### 4.1 Первая интерполяционная формула Ньютона (для интерполяции в начале таблицы)

Предположим, что точка $x$, в которой требуется найти значение функции, находится вблизи $x_0$, т.е. $q$ достаточно мало (обычно $0 < q < 1$). Тогда интерполяционный многочлен можно записать в виде:

$$
P_n(x_0 + qh) = y_0 + q\Delta y_0 + \frac{q(q-1)}{2!}\Delta^2 y_0 + \frac{q(q-1)(q-2)}{3!}\Delta^3 y_0 + \ldots
$$

$$
\ldots + \frac{q(q-1)\cdots(q-n+1)}{n!}\Delta^n y_0. \quad (8.26)
$$

Эта формула называется **первой интерполяционной формулой Ньютона** или формулой Ньютона для интерполяции **вперёд**.

**Обоснование.** Формула (8.26) получается последовательным применением оператора конечной разности к многочлену, записанному в базисе

$$
1, \quad (x - x_0), \quad (x - x_0)(x - x_1), \quad \ldots, \quad (x - x_0)(x - x_1)\cdots(x - x_{n-1}).
$$

Коэффициенты при этих базисных многочленах находятся через конечные разности в точке $x_0$.

**Замечание.** Первая формула Ньютона удобна для интерполяции в начале таблицы, когда $q$ находится в интервале $[0, 1]$. При $q > 1$ формула даёт экстраполяцию вправо, а при $q < 0$ — экстраполяцию влево, однако точность при этом падает.

### 4.2 Вторая интерполяционная формула Ньютона (для интерполяции в конце таблицы)

Если точка $x$ находится вблизи $x_n$ (т.е. в конце таблицы), то удобнее использовать вторую формулу Ньютона. В этом случае полагаем $x = x_n + qh$, где $q$ обычно удовлетворяет условию $-1 < q < 0$. Тогда интерполяционный многочлен имеет вид:

$$
P_n(x_n + qh) = y_n + q\Delta y_{n-1} + \frac{q(q+1)}{2!}\Delta^2 y_{n-2} + \frac{q(q+1)(q+2)}{3!}\Delta^3 y_{n-3} + \ldots
$$

$$
\ldots + \frac{q(q+1)\cdots(q+n-1)}{n!}\Delta^n y_0. \quad (8.28)
$$
Эта формула называется **второй интерполяционной формулой Ньютона** или формулой Ньютона для интерполяции **назад**.

**Обоснование.** Вторая формула Ньютона получается аналогично первой, но с использованием конечных разностей, взятых в окрестности точки $x_n$. Для этого используется базис

$$
1, \quad (x - x_n), \quad (x - x_n)(x - x_{n-1}), \quad \ldots, \quad (x - x_n)(x - x_{n-1})\cdots(x - x_1).
$$

Коэффициенты выражаются через конечные разности, расположенные на диагонали таблицы, ведущей к $y_n$.

**Замечание.** Вторая формула Ньютона особенно удобна, когда требуется вычислить значение функции в точке, близкой к правому концу таблицы, например, при экстраполяции вправо.



### 4.3 Сравнение формул Ньютона

Обе формулы Ньютона дают один и тот же многочлен, так как интерполяционный многочлен единственен. Выбор конкретной формулы определяется положением точки $x$ относительно узлов таблицы:

- Если $x$ находится ближе к $x_0$, используют первую формулу.
- Если $x$ находится ближе к $x_n$, используют вторую формулу.

Это позволяет уменьшить погрешность вычислений, связанную с накоплением ошибок округления при использовании разностей высоких порядков.


### 4.4 Оценка погрешности

Остаточный член для первой формулы Ньютона имеет вид:

$$
R_n(x) = \frac{h^{n+1}}{(n+1)!} q(q-1)\cdots(q-n) f^{(n+1)}(\xi),
$$

где $\xi$ — некоторая точка из интервала, содержащего все узлы и точку $x$.



Для второй формулы Ньютона остаточный член записывается аналогично:

$$
R_n(x) = \frac{h^{n+1}}{(n+1)!} q(q+1)\cdots(q+n) f^{(n+1)}(\xi).
$$

На практике для оценки погрешности часто используют приближённое равенство:

$$
|R_n(x)| \approx \left| \frac{q(q-1)\cdots(q-n)}{(n+1)!} \Delta^{n+1} y_0 \right|,
$$

если разности высших порядков меняются медленно.

### 4.5 Пример применения формул Ньютона

**Пример.** Для функции $y = \ln x$ построить интерполяционный многочлен Ньютона по узлам $x_0 = 2, x_1 = 3, x_2 = 4, x_3 = 5$ и найти приближённое значение $\ln(2.5)$.

**Решение.** Значения функции и конечные разности были вычислены в примере из § 8.4:

| $x$ | $y$ | $\Delta y$ | $\Delta^2 y$ | $\Delta^3 y$ |
|-----|-----|------------|--------------|--------------|
| 2   | 0.6931 | 0.4055 | -0.1178 | 0.0266 |
| 3   | 1.0986 | 0.2877 | -0.0646 | |
| 4   | 1.3863 | 0.2231 | | |
| 5   | 1.6094 | | | |



Для точки $x = 2.5$ имеем $q = \frac{2.5 - 2}{1} = 0.5$. Так как $x$ находится в начале таблицы, используем первую формулу Ньютона (8.26):

$$
P_3(2.5) = 0.6931 + 0.5 \cdot 0.4055 + \frac{0.5 \cdot (-0.5)}{2} \cdot (-0.1178) + \frac{0.5 \cdot (-0.5) \cdot (-1.5)}{6} \cdot 0.0266
$$

Выполним вычисления пошагово:

1. $0.6931 + 0.20275 = 0.89585$
2. $\frac{0.5 \cdot (-0.5)}{2} = -0.125$; $-0.125 \cdot (-0.1178) = 0.014725$
3. $0.89585 + 0.014725 = 0.910575$
4. $\frac{0.5 \cdot (-0.5) \cdot (-1.5)}{6} = \frac{0.375}{6} = 0.0625$; $0.0625 \cdot 0.0266 = 0.0016625$
5. $0.910575 + 0.0016625 = 0.9122375$

Таким образом, $P_3(2.5) \approx 0.91224$.

Точное значение $\ln(2.5) = 0.91629$. Абсолютная погрешность составляет $|0.91629 - 0.91224| = 0.00405$, что вполне приемлемо.
